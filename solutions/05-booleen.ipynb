On peut travailler aussi avec les **opérateurs logiques**.

`and` → opérateur logique "ET"

👉 Les deux conditions doivent être vraies


`or` → opérateur logique "OU"

👉 Une seule des conditions doit être vraie


`not` → opérateur logique "NON"

👉 Inverse le résultat

```python
print(True and True)
print(True and False)
```

```python
print(True or False)
print(False or False)
```

```python
print(not True) 
```
