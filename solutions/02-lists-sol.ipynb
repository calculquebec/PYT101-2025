# Lists

## Objective : learn to use lists

Lists are represented by square brackets `[]` and can contain any data structure, seperated by commas.

```python
[1, 2, 3, 4, 5]
```

A list can have multiple data types and even other lists as elements!

```python
[6, 'string', 7, 'another string']
```

```python
[True, ['another', 'list'], 8, 9, 10]
```

We can extract different parts of a list thanks to indexing. This works by adding a pair of brackets after the list surrounding the **index** of the element that we like to obtain. Note that in Python all indices start with a 0 — so in reality the first element is the element with index 0 (which is different from languages like R or MATLAB).

  <img src="fig/liste.png" alt="Photo centrée" width="600">

```python
my_list = ['red','green','blue','black','yellow','orange']
```

The element 'green' is the second element of `my_list`, so its index is `1`.

```python
my_list[1]
```

Note that we can also obtain a *slice* of values with the help of the colon operator (`:`).
A sole colon means to copy all elements.


```python
my_list[:]
```

A colon right of an index means that index and all that comes after.

```python
my_list[3:]
```

A colon left of an index mean everything before but **not** including that index.

```python
my_list[:3]
```

It is also possible to get a slice with indices for the start and end. Importantly, the end index is always excluded.

```python
my_list[1:4]
```

And using a negative index means that you take the elements starting at the end of the list, going towards the start. Note that the last element of the list does not use index `-0` but uses `-1`. The joys of Python!

```python
my_list[-1]
```

In the same way, you can obtain a slice using negative indices.

```python
my_list[-3:-1]
```

If we use an index outside the limits of the list, an error message is given.

```python
my_list[25]
```

### Manipulating lists

Here are some useful commands to manipulate lists.

- `len()` : returns the length of a list

```python
len(my_list)
```

- `.append()` : add an element to the end of the list

```python
my_list.append('purple')
```

```python
my_list
```

- combine lists. Importantly the `+` here does not let you add the numbers in the list, but rather concatenates the lists, similar to strings.

```python
second_list = ['bee', 'butterfly', 'dragonfly', 'fly']
second_list + my_list
```

```python
[10, 20, 30, 40, 50] + [1, 1, 1, 1, 1]
```

and not `[11, 21, 31, 41, 51]`

- `.count()` : to count the amount of times an element occurs in a list.

```python
latitude = [40.741895, 35.67686, 41.382894, 41.382894, 48.85889, 35.67686, 35.67686, 40.741895, 48.85889, 48.85889, 35.67686]
latitude.count(35.67686)
```

- `.sort()`: to put the elements in a list in order (depending on the data type). No output, modifies the list. Use `sorted()` instead if you want to keep the original.

```python
latitude.sort()
```

```python
latitude
```

- `set()` : returns unique values of a list

```python
set(latitude)
```
