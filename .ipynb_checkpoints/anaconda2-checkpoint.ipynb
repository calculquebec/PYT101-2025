# Anaconda

**Anaconda** est une distribution libre et open source des langages de programmation Python et R.
Elle vise à simplifier la gestion des paquets et le déploiement d'environnements de travail.
Anaconda est particulièrement populaire en science des données, apprentissage automatique et calcul scientifique.

Anaconda inclut notamment :
- **conda** : un gestionnaire de paquets et d'environnements
- **Anaconda Navigator** : une interface graphique pour gérer vos environnements et lancer des applications
- **JupyterLab / Jupyter Notebook** : pour créer des notebooks interactifs comme celui-ci
- Plus de 1500 paquets scientifiques préinstallés (NumPy, pandas, matplotlib, etc.)

> **Note :** Si vous avez aimé la convivialité de Jupyter pendant cette formation et souhaitez poursuivre votre apprentissage de Python sur votre propre ordinateur, Anaconda est une excellente façon de démarrer.

## 1. Installation d'Anaconda

### Étapes d'installation

1. Aller sur le site officiel : https://www.anaconda.com/download/success
2. Télécharger la version correspondant à votre système d'exploitation (**Windows**, **macOS** ou **Linux**)
3. Lancer l'installateur et suivre les étapes (vous pouvez garder tous les paramètres par défaut)
4. Une fois installé, lancer **Anaconda Navigator** depuis votre menu d'applications

### Vérification de l'installation

Après l'installation, vous pouvez vérifier que conda est disponible en ouvrant un terminal (ou l'**Anaconda Prompt** sous Windows) et en tapant :

```bash
conda --version
```

Vous devriez voir quelque chose comme :
```
conda 24.x.x
```

## 2. Anaconda Navigator

**Anaconda Navigator** est l'interface graphique d'Anaconda. Elle vous permet de :
- Lancer des applications comme JupyterLab, VS Code, Spyder, etc.
- Gérer vos environnements virtuels
- Installer, mettre à jour et supprimer des paquets

### Lancer JupyterLab depuis Navigator

1. Ouvrir **Anaconda Navigator**
2. Dans l'onglet **Home**, repérer la carte **JupyterLab**
3. Cliquer sur le bouton **Launch**

Cela ouvrira JupyterLab dans votre navigateur web par défaut.
Ce n'est **pas** parce que ça nécessite une connexion internet — les navigateurs sont simplement parfaits pour afficher des interfaces interactives.
Toutes vos données restent locales sur votre ordinateur.

### Naviguer dans JupyterLab

Une fois JupyterLab ouvert, vous verrez :
- Un **navigateur de fichiers** sur la gauche pour ouvrir vos notebooks (`.ipynb`)
- Un **espace de travail** au centre pour éditer et exécuter vos notebooks
- Une barre de menus en haut

## 3. Les environnements conda

Un **environnement** est un espace isolé contenant une version spécifique de Python et un ensemble de paquets.
Travailler avec des environnements permet d'éviter les conflits entre les dépendances de différents projets.

### Créer un environnement via Anaconda Navigator

1. Dans Navigator, cliquer sur **Environments** dans le menu de gauche
2. Cliquer sur le bouton **Create** (en bas de la liste)
3. Donner un nom à l'environnement (ex. : `mon-projet`)
4. Choisir la version de Python souhaitée
5. Cliquer sur **Create**

### Créer un environnement via le terminal

```bash
# Créer un environnement nommé "mon-projet" avec Python 3.11
conda create --name mon-projet python=3.11

# Activer l'environnement
conda activate mon-projet

# Désactiver l'environnement (retourner à base)
conda deactivate
```

### Lister les environnements existants

```bash
conda env list
```

## 4. Installer des paquets

### Via Anaconda Navigator (interface graphique)

1. Dans Navigator, cliquer sur **Environments**
2. Sélectionner l'environnement dans lequel installer le paquet
3. Dans le menu déroulant, choisir **Not installed** pour voir les paquets disponibles
4. Chercher le nom du paquet dans la barre de recherche
5. Cocher la case à côté du paquet souhaité
6. Cliquer sur **Apply** pour confirmer l'installation

> **Note :** Si le paquet est introuvable via l'interface graphique, il faut passer par la ligne de commande.

### Via conda (ligne de commande)

```bash
# Installer un paquet depuis les dépôts conda
conda install numpy

# Installer un paquet depuis un canal spécifique (ex. conda-forge)
conda install --channel conda-forge xarray

# Installer plusieurs paquets en même temps
conda install numpy pandas matplotlib
```

### Via pip (ligne de commande)

Si un paquet n'est pas disponible via conda, vous pouvez utiliser `pip` :

```bash
pip install nom-du-paquet
```

> **Conseil :** Préférez toujours `conda install` à `pip install` lorsque possible, pour éviter les conflits de dépendances.

## 6. Référence rapide des commandes conda

Voici un récapitulatif des commandes conda les plus utiles :

### Gestion des environnements

```bash
conda env list                          # Lister tous les environnements
conda create --name monenv python=3.11  # Créer un nouvel environnement
conda activate monenv                   # Activer un environnement
conda deactivate                        # Désactiver l'environnement actif
conda env remove --name monenv          # Supprimer un environnement
```

### Gestion des paquets

```bash
conda list                    # Lister les paquets installés
conda search numpy            # Chercher un paquet
conda install numpy           # Installer un paquet
conda update numpy            # Mettre à jour un paquet
conda remove numpy            # Désinstaller un paquet
conda update conda            # Mettre à jour conda lui-même
```

### Partager un environnement

```bash
# Exporter l'environnement dans un fichier
conda env export > environment.yml

# Recréer un environnement depuis un fichier
conda env create --file environment.yml
```
