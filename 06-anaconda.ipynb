**Anaconda** est une distribution libre et open source des langages de programmation Python et R.
Elle vise à simplifier la gestion des paquets et le déploiement d'environnements de travail.
Anaconda est particulièrement populaire en science des données, apprentissage automatique et calcul scientifique.

Anaconda inclut notamment :
- **conda** : un gestionnaire de paquets et d'environnements
- **Anaconda Navigator** : une interface graphique pour gérer vos environnements et lancer des applications
- **JupyterLab / Jupyter Notebook** : pour créer des notebooks interactifs comme celui-ci
- Plus de 1500 paquets scientifiques préinstallés (NumPy, pandas, matplotlib, etc.)

> **Note :** Si vous avez aimé la convivialité de Jupyter pendant cette formation et souhaitez poursuivre votre apprentissage de Python sur votre propre ordinateur, Anaconda est une excellente façon de démarrer.

## 1. Installation d'Anaconda

1. Aller sur le site officiel : https://www.anaconda.com/download/success
2. Télécharger la version correspondant à votre système d'exploitation (**Windows**, **macOS** ou **Linux**)
3. Lancer l'installateur et suivre les étapes (vous pouvez garder tous les paramètres par défaut)
4. Une fois installé, lancer **Anaconda Navigator** depuis votre menu d'applications

## 2. Anaconda Navigator

**Anaconda Navigator** est l'interface graphique d'Anaconda. Elle vous permet de :
- Lancer des applications comme JupyterLab, VS Code, Spyder, etc.
- Gérer vos environnements virtuels
- Installer, mettre à jour et supprimer des paquets

### Lancer JupyterLab depuis Navigator

1. Ouvrir **Anaconda Navigator**
2. Dans l'onglet **Home**, repérer la carte **JupyterLab**
3. Cliquer sur le bouton **Launch**

Cela ouvrira JupyterLab dans votre navigateur web par défaut.
Ce n'est **pas** parce que ça nécessite une connexion internet — les navigateurs sont simplement parfaits pour afficher des interfaces interactives.
Toutes vos données restent locales sur votre ordinateur.

### Naviguer dans JupyterLab

Une fois JupyterLab ouvert, vous verrez :
- Un **navigateur de fichiers** sur la gauche pour ouvrir vos notebooks (`.ipynb`)
- Un **espace de travail** au centre pour éditer et exécuter vos notebooks
- Une barre de menus en haut

## 3. Installer des paquets

1. Dans Navigator, cliquer sur **Environments**
2. Sélectionner l'environnement dans lequel installer le paquet
3. Dans le menu déroulant, choisir **Not installed** pour voir les paquets disponibles
4. Chercher le nom du paquet dans la barre de recherche
5. Cocher la case à côté du paquet souhaité
6. Cliquer sur **Apply** pour confirmer l'installation

> **Note :** Si le paquet est introuvable dans la barre de recherche, son installation exige de passer par la ligne de commande. Les installations avancées seront présentées dans PYT102 Perfectionnement en programmation Python.

## 5. Environnements de développement pour Python

Il existe plusieurs façons d'écrire et d'exécuter du code Python. Voici un aperçu des environnements les plus courants :

| Environnement | Type | Points forts | Idéal pour |
|---------------|------|--------------|------------|
| **JupyterLab** | Navigateur web | Cellules interactives, visualisations intégrées, format `.ipynb` | Science des données, enseignement, exploration |
| **Spyder** | IDE | Interface familière (comme MATLAB), explorateur de variables | Calcul scientifique, débutants |
| **VS Code** | Éditeur de code | Très extensible, supporte plusieurs langages, débogueur puissant | Développement général, projets complexes |
| **PyCharm** | IDE | Complétion de code avancée, refactorisation, outils intégrés | Développement Python professionnel |
| **Google Colab** | Navigateur web (cloud) | Aucune installation, accès GPU gratuit, partage facile | Apprentissage automatique, collaboration |

**JupyterLab** et **Spyder** sont inclus avec Anaconda et peuvent être lancés directement depuis Anaconda Navigator.

**VS Code** peut aussi être lancé depuis Navigator si vous l'avez installé au préalable.
