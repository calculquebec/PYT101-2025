### Mathematical operations

The most basic use of Python is to use it like a sophisticated calculator. It is very easy to do simple calculations in Python.

```python
print(5 + 1)
```

Note that in a notebook, unlike in a script, we don't always have to use the `print()` instruction. Not using `print()` gives us the same result as before.

```python
5 + 1
```

Python can do all the mathematical operations that you would expect.

```python
5 + 3
```

```python
2 - 9
```

```python
4 * 6
```

```python
14 / 3
```

You can use it to do more complex operations, such as exponentials with `**`.

```python
5 ** 2
```

For integer division the `//` operator gives the quotient rounded down. The modulo operator `%` gives the remainder after division.

```python
5 // 2
```

```python
5 % 2
```

### Variables 

Of course, we eventually would like to save our results. To do this, we can use variables. In Python, a variable is a name given to a saved result. We can define one with the equal sign `=`.

```python
weight_kg = 52
```

If we want to retrieve the information we saved, we just need to retype the variable name.

```python
weight_kg
```

We can do mathematical operations on variables in the same way as we would normally do.

```python
print('Poids en lb:', 2.2 * weight_kg)
```

Like the above example show, we can show multiple values at the same time by separating them using commas. We can also change the value of a variable by assigning it a new one:

```python
weight_lb = 2.2 * weight_kg
print(weight_lb)
```

What happens when we change a variable ? Let's change `weight_kg` and see what happens to `weight_lb`.

```python
weight_kg = 57
print(weight_lb)
```

Note that even though we changed the value of `weight_kg`, `weight_lb` is not automatically updated. This shows a very important property of most programming languages: a computer does nothing unless we specifically tell it to do something, things don't happen automatically. This differs from the behaviour of a spreadsheet where a cell is automatically updated when it refers to another cell that has been modified.
If we want to tell Python to update `weight_lb` to reflect the new value of `weight_kg`, we need to perform this operation explicitly.

```python
weight_lb = weight_kg * 2.2
print('The new value of weight_lb is', weight_lb)
```

### Spaces or no spaces ?

In general, spaces can be used between operators (`+, -, =, *`) and after commas to improve visibility:

`x = 3 + 4` vs `x=3+4`

`print(1, 2, 3)` vs `print(1,2,3)`

On the other hand, using spaces between a function and its parentheses is not recommended, and putting spaces in variable names gives rise to error messages.

`print ("Not recommended")` vs `print("Much better !")`

`a variable name = "that causes error messages"'` vs `a_good_name = "without errors"`

### Data types

Variables are not only used to save numerical values. Other data types exist.

- Character sequences (*strings*). Strings are used to store text. They are put between single (`' '`) or double (`" "`) quotes.

```python
print('this is a string')
print("this is also a string")
```

Python makes it very easy to work with simple text. For example, we can even use the `+` sign to concatenate strings, or `*` to repeat them multiple times.

```python
'This is a string' + ', that I can extend'
```

```python
'repetition' * 3
```

Note that you can't do arbitrary mathematical operations on strings. For example, you cannot add a number to string!

```python
'5' + 5
```

The error message shows us that it's not possible to concatenate a number to a string. Those are two different data types. If you want to know the data type of a variable you can use the `type()` function.

```python
type('string')
```

str for *string*

```python
type(5)
```

```python
type(5.4)
```

### Going further

Multiple useful methods can be applied to strings. Here are some that could be practical in your projects. For a more exhaustive list please see: https://www.w3schools.com/python/python_ref_string.asp

`count()` : returns the number of times that a specific value appears in a string

```python
butterfly = 'I prefer orange butterflies and yellow butterflies'
```

```python
butterfly.count('butterflies')
```

`join()`: combine elements of a string tuple, with the specified seperator

```python
'-'.join(('2025','05','24'))
```

`split()` : the inverse of `join()`, separate the string in substrings in a list, using the specified separator.

```python
'2025-05-24'.split('-')
```

`startswith()` : returns `True` if the string starts with the specified prefix, otherwise `False`

```python
butterfly.startswith('The')
```

Other more generic functions can be applied to strings as well.

`len()` : returns the length of the string

```python
len(papillon)
```

`set()` : returns unique elements

```python
sequence = 'AAGGTTTCCAGCTAGGTACCCGATATTAGTTTCCAAAATTGCGA'
```

```python
set(sequence)
```
