### Scripts

Python programs (scripts) are simply a collection of Python commands saved in a file. No need to compile! To show this, let's write our first program!

Copy the following text into a text editor and save it using a filename of your choice (Python files usually use the extension .py).

`print("I just ran my first python script!!!")`

We can now run this program in multiple ways. If we open a terminal in the folder where we saved our program, we can then use the following command to run it:

`python my first_script.py`

Python notebooks also allow you to run Bash commands inside the notebook. You just need to add a `!` before the command to run.

`!python my_first_script.py`

### Importing Python packages

Python includes numerous supplementary features but not all are loaded by default. To use a package, we have to import it. Some of those packages are general, where others are for very specific uses cases. Here are some popular packages:
- pandas
- numpy
- matplotlib
- os
- seaborn
- scikit-learn

To stay within the spirit of our script, we will test the import of the `sys` package which allows us to obtain the command line arguments.

```python
import sys
```

To test the `sys` functionality, copy the following text in a text editor and save it using a name of your choosing.

`import sys` 

`print('We wrote: ', sys.argv)`

Run the script and add some words or numbers after the script name, for example:

`!python test_sys.py something to test this package 1 2 3 4 5`

Output: 

`We wrote: ['test_sys.py', 'something', 'to', 'test', 'this', 'package', '1', '2', '3', '4', '5']`

The `sys.argv` variable is a list which contains the arguments that were passed to the script using the command line.

The `sys` package can be very useful when a script must be used multiple times with different input files. You can even specify variables on the command line, without needing to change the script.

### What if the package is missing?

You may try to import a package but could get the following error:

```python
import a_nonexisting_package
```

The first thing to do is to make sure you didn't make a typo. If the error persists, the package is probably missing on your computer and must be downloaded before it can be imported. ### Anaconda

If you start programming, an easy method is to use Anaconda. Anaconda is an open source distribution of Python and R which aims to simplify the management of packages and their use. Anaconda also lets you start Jupyter notebooks, like those used in this workshop. If you liked the way Jupyter worked during this workshop and you would like to continue to learn Python with this interactive tool, here is how to istall it.

To install Anaconda :
- Go to the site https://www.anaconda.com/download/success
- Download the version for your system (Windows, macOS, Linux)
- Install Anaconda (just "Next", nothing to configure)

To installe a package from PyPi:
- Click on Environments in the left menu
- Look for the name of the package in the search bar
- Select the name of the package and click on Apply
- If the package can't be found in the search bar, you may need to install it using the command line. Advanced installations are presented in PYT102 Enhancing your Python programming skills.

To start a Notebook :
- Open Anaconda Navigator
- Launch JupyterLab via the Navigator using the "Launch" button. This will open a web interface (in your default browser), not because it needs to use the internet but because browsers are perfect tools for show interactive interfaces.
- After that you can open notebooks (extension .ipynb) using the file opening dialog box or by creating a new one.
