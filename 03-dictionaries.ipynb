# Dictionaries

## Objective : learn how to use dictionaries

Dictionaries are another kind of data structure in Python. They are very useful, because they allow us to rapidly store and retrieve **key-value** pairs. Dictionaries are defined using curly brackets `{}`. They are a bit like a real dictionary: you look for a word (the *key*) to get its definition (*value*).

Example: length in metres of our favourite marine mammals:

`marine_mammal_length = {'rorqual' : 20, 'narval' : 5, 'beluga' : 4}`

In this example, rorqual, narval and beluga are the **keys**, whereas 20, 5 and 4 are the **values**. Like in a real dictionary, **each key must be unique**, whereas a value can be the same for multiple keys.

- **Declaring a dictionary** : to create a new dictionary, you can first create an emty dictionary and then fill it, or you can create and fill it in the same command.

```python
capitals = {}
```

```python
capitals
```

```python
type(capitals)
```

```python
capitals['France'] = 'Paris'
capitals['Canada'] = 'Ottawa'
capitals['Italia'] = 'Rome'
capitals['Spain'] = 'Madrid'
```

```python
capitals
```

```python
elements = {"Hydrogen": "H", "Oxygen": "O", "Carbon": "C", "Nitrogen": "N"}
```

```python
elements
```

### Useful commands

```python
capitals.keys()
```

- `.values()` : to get all values for a dictionary
